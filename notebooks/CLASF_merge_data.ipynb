este notebook es simplemente para juntas los datasets

```python
import pandas as pd

datasets = ['CLUST_Digital_Music.jsonl','CLUST_Gift_Cards.jsonl',
            'CLUST_Magazine_Subscriptions.jsonl','CLUST_Subscription_Boxes.jsonl']

categories = ['digital_music','gift_cards','magazines','subscription_boxes']


dfs = []
for i,j in zip(datasets,categories):

    path = f'../data/raw/{i}'
    temp_df = pd.read_json(path,lines=True)
    temp_df['y'] = j
    dfs.append(temp_df)
final_df = pd.concat(dfs)
final_df.reset_index()
final_df.to_parquet('../data/interim/CLUST_data.parquet')
```
