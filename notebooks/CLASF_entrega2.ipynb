```python
import umap
from sklearn.manifold import TSNE
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
from sklearn.metrics import silhouette_score

seed = 89

colors = ["#e6e5e6","#3b5247","#932625"]
colormap = mpl.colors.LinearSegmentedColormap.from_list("colorcitos",colors)
colorpalette = colormap(np.linspace(-1,1,10))[::-1]
sns.set_theme(style='white',font='DejaVu Serif',palette=colorpalette)



inmutable_df = pd.read_parquet('../data/processed/CLASF_data.parquet')
```

```python
df = inmutable_df.copy()


df
```

```python

X = df.iloc[:,8:]
Y = df['y']

```

# T-sne

```python

small_X = df.iloc[:,7:].sample(n=10_000)
small_Y = small_X['y']
small_X = small_X.drop(columns='y')
```

```python
# Continuemos con la exploracion

# tomememos un subconjunto de los datos



for p in range(10,160,10):
    X_embedded = TSNE(n_components=2, learning_rate='auto',init='random', perplexity=p,random_state=seed).fit_transform(small_X)
    sns.scatterplot(x=X_embedded[:,0],y=X_embedded[:,1],hue=small_Y,palette=colorpalette[:5].tolist())
    silhouette = silhouette_score(X_embedded,small_Y)
    plt.title(f'perplexity = {p}\n silhouette = {silhouette}')
    plt.show()
```

# UMAP


pese a que de momento  solo estamos realizando estas proyecciones con afan de entender si hay clusters claros

```python
# Vamos haciendo el embedding basico naive que umap nos propone, despues tuneamos hiperparametros
umap_tranformer = umap.UMAP()

umap_data = umap_tranformer.fit_transform(small_X)
sns.scatterplot(x= umap_data[:,0],y=umap_data[:,1],hue=small_Y,alpha=.4)
silhouette = silhouette_score(umap_data,small_Y)
plt.title(f'Uniform Manifold Aproximation and Projection\nSilhouette = {silhouette}')
```

veamos si podemos mejorarlo tuneando hiperparametros

el principal hiperparametro del UMAP es $K$ de $K$-nearest neighbors, para entender mejor como funciona UMAP puede consultar las referencias o las notas que _probablemente_ haga y deje en el directorio `docs`


aunque de primeras lo que de primeras se aprecia tanto de UMAP y de t-SNE es que los clusters estan bastante superpuestos y que `subscription_boxes` es la clase mas superpuesta y dificil de separar

### Desventajas de UMAP y t-SNE

La principal desventaja o problema que creo que podriamos encontrar al trabajar con los datos con los que estamos trabajando es que los dos ambos algoritmos de reduccion de dimensionalidad se basa principalmente en la distancia euclidiana de un punto a otro, evidentemente hay otras cosas en consideracion pero las loss function de ambos dependen en gran manera de la distancia euclidiana. Y como ya mencionamos antes, la gracia de hacer sentence embedding es que la semantica se encodea principalmente atraves de las distancias coseno de un vector a otro, es decir, es probable que un algoritmo de reduccion de dimensionalidad que tome en cuaenta esto podria llegar a ser mas adecuado o ayudaria a mantener la estructura que buscamos. esto podria ser interesante.
